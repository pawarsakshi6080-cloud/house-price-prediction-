## Data Preprocessing 

```python
X = train.drop('SalePrice', axis=1)
y = train['SalePrice']
```

```python
print("X shape:",X.shape)
print("y shape:",y.shape)
```

```python
X = X.drop('Id',axis=1)
test = test.drop('Id', axis=1) 
```

```python
missing_percentage = (X.isnull().sum()/len(X))*100 
missing_percentage = missing_percentage[missing_percentage > 0].sort_values(ascending=False)

missing_percentage 
```

```python
missing_summary = pd.DataFrame({'Missing_Count':                                # To make a missing value summary 
                                X.isnull().sum(),'Missing_Percentage' :
                                (X.isnull().sum() / len(X))*100,
                                'Unique_Values': X.nunique(),
                                'Data_Type': X.dtypes 
})

missing_summary = missing_summary[missing_summary['Missing_Count']>0].sort_values('Missing_Percentage',ascending= False)
missing_summary 
```

```python
categorical_missing = X.select_dtypes(include='object').columns[             # To separate categorical and numerical missing clumns 
    X.select_dtypes(include='object').isnull().any()
]
numerical_missing = X.select_dtypes(exclude='object').columns[
X.select_dtypes(exclude='object').isnull().any()
]
print("Categorical clumns with missing values:")
print(categorical_missing)

print("\nNumerical columns with missing values:")
print(numerical_missing)
```

```python
for col in categorical_missing:
    print("\n", col)
    print(X[col].value_counts(dropna=False))
```

```python
# to drop columns with extremely high missing values 

drop_cols = ['PoolQC','MiscFeature','Alley','Fence']
X = X.drop(columns = drop_cols)
test = test.drop(columns= drop_cols)
```

```python
# to fill categorical columns where NaN means "feature doesn't exist"

none_cols = ['MasVnrType','FireplaceQu','GarageType','GarageFinish','GarageQual','GarageCond','BsmtQual', 'BsmtCond','BsmtExposure',
       'BsmtFinType1', 'BsmtFinType2']
for col in none_cols:
    X[col] = X[col].fillna('None')
    test[col] = test[col].fillna('None')
```

```python
# to fill numerical columns representing absence 

zero_col = ['MasVnrArea', 'GarageYrBlt']
for col in zero_col: 
    X[col] = X[col].fillna(0)
    test[col] = test[col].fillna(0)
```

```python
# LotFrontage depends strongly on the neighborhood so used the median of each neighborhood 

X['LotFrontage'] = X.groupby('Neighborhood')['LotFrontage'].transform(lambda X: X.fillna(X.median()))
test['LotFrontage'] = test.groupby('Neighborhood')['LotFrontage'].transform(lambda X: X.fillna(X.median()))
```

```python
# remaning categorical column Electrical has only 1 missing value so filled with most common category 

X['Electrical'] = X['Electrical'].fillna(X['Electrical'].mode()[0])
test['Electrical'].fillna(X['Electrical'].mode()[0])
```

```python
# to check filled values 
missing_after = X.isnull().sum()
missing_after = missing_after[missing_after > 0].sort_values(ascending=False)
missing_after 
```

```python
# column that should not be treated as continuous numerical features 
special_numeric_cols = ['MSSubClass','MoSold']

# Continuous/quantitative columns that will be scaled 
scale_continuous_cols = [
    col for col in continuous_cols
    if col not in 
special_numeric_cols
 ]

# final columns for standardScaler 
scaling_cols = ordinal_cols + scale_continuous_cols

print("Number of columns to scale:",len(scaling_cols))
print("\nScaling Columns:")
for col in scaling_cols:
    print(col)
```

```python
# Make sure test data has exactly the same feature columns as training data 
X_test = test 
X_test = X_test.reindex(columns=X.columns, fill_value = 0) 
print("X shape:",X.shape)
print("X_test shape:", X_test.shape)
```

### Spliting tha data 

```python
# split ONLY the training data 
from sklearn.model_selection import train_test_split
X_train, X_val, y_train, y_val = train_test_split( X, y, test_size = 0.20, random_state = 42)

print("X_train shape:", X_train.shape)
print("X_val shape:",X_val.shape)
print("y_train shape:",y_train.shape)
print("y_val shape:",y_val.shape)
```

### Scaling

```python
# Apply StandardScaler 
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Make copies so original datasets remain unchanged 
X_train_scaled = X_train.copy()
X_val_scaled = X_val.copy()
X_test_scaled = X_test.copy()

# Fit ONLY on training data 
X_train_scaled[scaling_cols] = scaler.fit_transform(X_train[scaling_cols])

# Use the same fitted scaler for validation and test 
X_val_scaled[scaling_cols] = scaler.transform(X_val[scaling_cols])
X_test_scaled[scaling_cols] = scaler.transform(X_test[scaling_cols])

print("Scaling completed successfully.")
```

```python
# check mean and standard deviation after scaling 
scaled_check = pd.DataFrame({
    'mean':
    X_train_scaled[scaling_cols].mean(),'std':
    X_train_scaled[scaling_cols].std(ddof = 0)
 })

print(scaled_check)
```

```python
print("Maximum absolute mean:", X_train_scaled[scaling_cols].mean().abs().max())
print("Maximum difference from std=1:",(X_train_scaled[scaling_cols].std(ddof=0) - 1 ).abs().max())
```

```python
# verify binary columns stayed 0/1 
binary_check = X_train_scaled[binary_cols].nunique() 
print("Maximum number of unique values in binary columns:",binary_check.max())
```
